```python
import pandas as pd

file_path = "../data/raw/ibtracs.NI.list.v04r01.csv"

real_cyclones = pd.read_csv(file_path, skiprows=[1])

print("Dataset loaded successfully!")
print("Rows:", len(real_cyclones))
print("Columns:", len(real_cyclones.columns))

real_cyclones.head()
```

```python
print(real_cyclones.columns.tolist())
```

```python
important_columns = [
    "SID",
    "SEASON",
    "NAME",
    "ISO_TIME",
    "LAT",
    "LON",
    "USA_WIND",
    "USA_PRES",
    "USA_SSHS",
    "STORM_SPEED",
    "STORM_DIR"
]

real_cyclones[important_columns].head(20)
```

```python
recent_cyclones = real_cyclones[
    (real_cyclones["SEASON"] >= 2000) &
    (real_cyclones["NAME"] != "UNNAMED")
]

recent_cyclones[[
    "SEASON",
    "NAME",
    "ISO_TIME",
    "LAT",
    "LON",
    "USA_WIND",
    "USA_PRES",
    "STORM_SPEED",
    "STORM_DIR"
]].head(20)
```

```python
cyclone_list = (
    recent_cyclones[["SEASON", "NAME"]]
    .drop_duplicates()
    .sort_values(["SEASON", "NAME"], ascending=[False, True])
)

cyclone_list.head(30)
```

```python
selected_cyclone = cyclone_list.iloc[0]

print("Selected Cyclone:")
print("Year:", selected_cyclone["SEASON"])
print("Name:", selected_cyclone["NAME"])
```

```python
selected_name = selected_cyclone["NAME"]
selected_year = selected_cyclone["SEASON"]

storm = recent_cyclones[
    (recent_cyclones["NAME"] == selected_name) &
    (recent_cyclones["SEASON"] == selected_year)
].copy()

storm = storm.sort_values("ISO_TIME")

print("Cyclone:", selected_name)
print("Year:", selected_year)
print("Observations:", len(storm))

storm[[
    "ISO_TIME",
    "LAT",
    "LON",
    "USA_WIND",
    "USA_PRES",
    "STORM_SPEED",
    "STORM_DIR"
]].head(20)
```

```python
dna_features = [
    "USA_WIND",
    "USA_PRES",
    "STORM_SPEED",
    "STORM_DIR",
    "LAT",
    "LON"
]

dna_data = storm[["ISO_TIME"] + dna_features].copy()

# Convert values to numbers
for column in dna_features:
    dna_data[column] = pd.to_numeric(
        dna_data[column],
        errors="coerce"
    )

# Remove observations where important values are missing
dna_data = dna_data.dropna(subset=dna_features)

print("Cyclone DNA observations:", len(dna_data))

dna_data.head(10)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

dna_values = scaler.fit_transform(dna_data[dna_features])

real_dna = pd.DataFrame(
    dna_values,
    columns=dna_features
)

real_dna.insert(0, "ISO_TIME", dna_data["ISO_TIME"].values)

print("Real Cyclone DNA created!")
real_dna.head(10)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Take the latest cyclone observation
latest_dna = real_dna.iloc[-1]

dna_labels = [
    "Wind",
    "Pressure",
    "Movement Speed",
    "Movement Direction",
    "Latitude",
    "Longitude"
]

dna_values = [
    latest_dna["USA_WIND"],
    latest_dna["USA_PRES"],
    latest_dna["STORM_SPEED"],
    latest_dna["STORM_DIR"],
    latest_dna["LAT"],
    latest_dna["LON"]
]

# Close the radar chart
angles = np.linspace(0, 2 * np.pi, len(dna_labels), endpoint=False).tolist()

dna_values += dna_values[:1]
angles += angles[:1]

# Create radar chart
fig, ax = plt.subplots(
    figsize=(8, 8),
    subplot_kw=dict(polar=True)
)

ax.plot(angles, dna_values, linewidth=2)
ax.fill(angles, dna_values, alpha=0.25)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(dna_labels)

ax.set_ylim(0, 1)

ax.set_title(
    f"🧬 Cyclone DNA Profile\n{selected_name} ({selected_year})",
    pad=25
)

plt.show()
```

```python
# Convert wind and pressure to numeric values
storm["USA_WIND"] = pd.to_numeric(storm["USA_WIND"], errors="coerce")
storm["USA_PRES"] = pd.to_numeric(storm["USA_PRES"], errors="coerce")

# Make sure observations are in time order
storm = storm.sort_values("ISO_TIME").copy()

# Calculate changes between consecutive observations
storm["WIND_CHANGE"] = storm["USA_WIND"].diff()
storm["PRESSURE_CHANGE"] = storm["USA_PRES"].diff()

# Approximate wind intensification rate per hour
storm["WIND_INTENSIFICATION"] = storm["WIND_CHANGE"] / 3

# Display the result
storm[[
    "ISO_TIME",
    "USA_WIND",
    "USA_PRES",
    "WIND_CHANGE",
    "PRESSURE_CHANGE",
    "WIND_INTENSIFICATION"
]].head(20)
```

```python
# Create an intensification score
storm["INTENSIFICATION_SCORE"] = (
    storm["WIND_INTENSIFICATION"]
    - storm["PRESSURE_CHANGE"] / 3
)

# Show the result
storm[[
    "ISO_TIME",
    "USA_WIND",
    "USA_PRES",
    "WIND_INTENSIFICATION",
    "PRESSURE_CHANGE",
    "INTENSIFICATION_SCORE"
]].head(20)
```

```python
# Features for our improved Cyclone DNA
dna_features_v2 = [
    "USA_WIND",
    "USA_PRES",
    "STORM_SPEED",
    "WIND_INTENSIFICATION",
    "PRESSURE_CHANGE",
    "INTENSIFICATION_SCORE"
]

# Keep only the required columns
dna_v2_data = storm[["ISO_TIME"] + dna_features_v2].copy()

# Remove rows with missing values
dna_v2_data = dna_v2_data.dropna(subset=dna_features_v2)

# Normalize all DNA features to 0–1
scaler_v2 = MinMaxScaler()

dna_v2_values = scaler_v2.fit_transform(
    dna_v2_data[dna_features_v2]
)

real_dna_v2 = pd.DataFrame(
    dna_v2_values,
    columns=dna_features_v2
)

real_dna_v2.insert(
    0,
    "ISO_TIME",
    dna_v2_data["ISO_TIME"].values
)

print("Improved Cyclone DNA created!")
print("Number of DNA observations:", len(real_dna_v2))

real_dna_v2.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

for feature in dna_features_v2:
    plt.plot(
        real_dna_v2["ISO_TIME"],
        real_dna_v2[feature],
        marker="o",
        label=feature
    )

plt.title(
    f"🧬 Cyclone DNA Evolution — {selected_name} ({selected_year})"
)

plt.xlabel("Time")
plt.ylabel("Normalized DNA Value (0–1)")

plt.xticks(rotation=45)
plt.legend(
    bbox_to_anchor=(1.05, 1),
    loc="upper left"
)

plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Classify cyclone behaviour at each observation

def classify_behaviour(row):
    if row["WIND_INTENSIFICATION"] > 0 and row["PRESSURE_CHANGE"] < 0:
        return "Strengthening"
    elif row["WIND_INTENSIFICATION"] < 0 and row["PRESSURE_CHANGE"] > 0:
        return "Weakening"
    else:
        return "Stable"

storm["BEHAVIOUR"] = storm.apply(classify_behaviour, axis=1)

storm[[
    "ISO_TIME",
    "USA_WIND",
    "USA_PRES",
    "WIND_INTENSIFICATION",
    "PRESSURE_CHANGE",
    "BEHAVIOUR"
]].head(30)
```

```python
# View the complete intensity evolution

lifecycle_data = storm[[
    "ISO_TIME",
    "USA_WIND",
    "USA_PRES",
    "WIND_INTENSIFICATION",
    "PRESSURE_CHANGE",
    "BEHAVIOUR"
]].copy()

lifecycle_data
```

```python
def lifecycle_stage(row):
    wind = row["USA_WIND"]
    pressure = row["USA_PRES"]
    behaviour = row["BEHAVIOUR"]

    if pd.isna(wind) or pd.isna(pressure):
        return "Unknown"

    if behaviour == "Strengthening":
        if wind >= 64:
            return "Rapid Intensification"
        else:
            return "Intensification"

    elif behaviour == "Weakening":
        return "Weakening"

    else:
        if wind >= 64:
            return "Mature"
        else:
            return "Organisation"


storm["LIFECYCLE_STAGE"] = storm.apply(
    lifecycle_stage,
    axis=1
)

storm[[
    "ISO_TIME",
    "USA_WIND",
    "USA_PRES",
    "BEHAVIOUR",
    "LIFECYCLE_STAGE"
]]
```

```python
# Convert lifecycle stages into numerical levels for visualization

stage_map = {
    "Organisation": 1,
    "Intensification": 2,
    "Rapid Intensification": 3,
    "Mature": 4,
    "Weakening": 5
}

storm["LIFECYCLE_VALUE"] = storm["LIFECYCLE_STAGE"].map(stage_map)

plt.figure(figsize=(14, 6))

plt.plot(
    storm["ISO_TIME"],
    storm["LIFECYCLE_VALUE"],
    marker="o",
    linewidth=2
)

plt.yticks(
    list(stage_map.values()),
    list(stage_map.keys())
)

plt.xlabel("Time")
plt.ylabel("Lifecycle Stage")

plt.title(
    f"🌪️ Cyclone Lifecycle — {selected_name} ({selected_year})"
)

plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```
